```python
# Arithmetic operators
a = 10
b = 3
print(f"a + b = {a + b}")
print(f"a - b = {a - b}")
print(f"a * b = {a * b}")
print(f"a / b = {a / b}")
print(f"a % b = {a % b}")
print(f"a // b = {a // b}")
print(f"a ** b = {a ** b}")

# Comparison operators
x = 10
y = 12
print(f"x > y is {x > y}")
print(f"x < y is {x < y}")
print(f"x == y is {x == y}")
print(f"x != y is {x != y}")
print(f"x >= y is {x >= y}")
print(f"x <= y is {x <= y}")

# Logical operators
p = True
q = False
print(f"p and q is {p and q}")
print(f"p or q is {p or q}")
print(f"not p is {not p}")

# Assignment operators
c = 10
c += 5
print(f"c after c += 5 is {c}")

# Bitwise operators
m = 10 # Binary: 1010
n = 4  # Binary: 0100
print(f"m & n = {m & n}")
print(f"m | n = {m | n}")
print(f"m ^ n = {m ^ n}")
print(f"~m = {~m}")
print(f"m << 2 = {m << 2}")
print(f"m >> 2 = {m >> 2}")
```

The `open(`) function in Python is used to open files. It takes several parameters, but the most common ones are file and mode.

file: This is a required parameter that specifies the path to the file you want to open.
mode: This is an optional parameter that specifies the mode in which you want to open the file. If you don't specify a mode, it defaults to` 'r'` (read mode).
Here are the most common modes:

`'r'` (Read): Opens a file for reading. The file pointer is at the beginning of the file. This is the default mode.

`'w'` (Write): Opens a file for writing. If the file already exists, its content is truncated (emptied). If the file does not exist, it creates a new file.

`'a'` (Append): Opens a file for appending. The file pointer is at the end of the file. If the file does not exist, it creates a new file.

`'x'` (Exclusive Creation): Creates a new file, but raises an error if the file already exists.

`'t'` (Text): Opens the file in text mode. This is the default mode.

`'b'` (Binary): Opens the file in binary mode.

`'+'` (Update): Opens a file for both reading and writing.
You can combine modes, for example:

`'r+'`: Opens a file for both reading and writing (the file pointer is at the beginning).
`'w+'`: Opens a file for both writing and reading (truncates the file if it exists, otherwise creates a new file).
`'a+'`: Opens a file for both appending and reading (the file pointer is at the end).
